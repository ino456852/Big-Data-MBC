```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
titanic_data = sns.load_dataset("titanic") 
titanic_data
```

```python
print(titanic_data.info())
print(titanic_data.nunique())
print(titanic_data.describe(include='object'))
```

```python
titanic_data.isna().sum() # 컬럼별 결측치 갯수
titanic1 = titanic_data.copy() # 복사
titanic1['age'] = titanic1['age'].fillna(titanic1['age'].mean()) # age의 결측치 age의 평균값으로 채움
titanic1.info()
```

```python
# 결측치 처리
# freq = titanic1['embarked'].value_counts().index[0]
# titanic1['embarked'] = titanic1['embarked'].fillna(freq) # 최빈값으로 대체
# 또는
titanic1['embarked'] = titanic1['embarked'].fillna(titanic1['embarked'].mode()[0])
titanic1['embarked']
```

```python
# 결측치가 너무 많은 컬럼 삭제 embarked town은 embarked와 같은 데이터 이므로 삭제
drop_col = ['deck','embark_town']
titanic1 = titanic1.drop(columns=drop_col)
titanic1.info()
```

```python
# traning data set과 test data set을 75:25 비율 분리
# random_state은 100, stratify 옵션 사용
x = titanic1.drop(['survived'],axis=1)
y = titanic1['survived']
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y, test_size=0.25,random_state=100,stratify=y)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# 필요한 Label Encoding/One-hot Encoding을 실시하시오
x_train.info()
# data의 자료형이 obejct,category,bool 인 경우
# Label Encoding또는 One-hot Encoding 실시

# Label Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
label_list = ['sex','embarked','who','adult_male','alive','alone'] # Label Encoding 필요 columns
x_train[label_list] = x_train[label_list].apply(le.fit_transform)
x_test[label_list] = x_test[label_list].apply(le.fit_transform)

# One-hot Encoding
x_train = pd.get_dummies(x_train)
x_test = pd.get_dummies(x_test)

# train과 컬럼 순서 동일화
x_test = x_test[x_train.columns]
```

```python
# 데이터가 숫자형인 경우 Scaling 필요
# Standard scaling 또는 MinMax scaling 필요

# StandardScaler
x_train.info()
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scale_list = ['pclass','age','sibsp','parch','fare']

scaler.fit(x_train[scale_list])

x_train[scale_list] = scaler.transform(x_train[scale_list])
x_test[scale_list] = scaler.transform(x_test[scale_list])

# MinMaxScaler
# from sklearn.preprocessing import MinMaxScaler
# scaler = MinMaxScaler()
# scale_list = ['pclass','age','sibsp','parch','fare']
# scaler.fit(x_train[scale_list]) # fit은 train 데이터만 적용

# x_train[scale_list] = scaler.transform(x_train[scale_list])
# x_test[scale_list] = scaler.transform(x_test[scale_list])
```

```python
# Encoding과 Scaler 동시에 실행

from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
check_num = x_train.select_dtypes(include=['number'])
print(check_num.columns)
check_obj = x_train.select_dtypes(include=['object','category','bool'])
print(check_obj.columns)

ct = make_column_transformer(
    (StandardScaler(), check_num.columns),
    (OneHotEncoder(sparse_output=False),check_obj.columns))

ct.fit(x_train)
x_train1 = ct.transform(x_train)
x_test1 = ct.transform(x_test)
```

```python
# RandomForestClassifier 사용 모델을 생성하고 학습 시키시오
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
# model = RandomForestClassifier(max_depth=5,n_estimators=300)
model.fit(x_train1,y_train)
```

```python
# 분리한 test data set을 이용하여 분류 예측을 실행하시오
# x_test data set에 대한 Y 예측값
Y_pred = model.predict(x_test1)
print(Y_pred)
```

```python
# 분류 예측의 accuracy, f1-score, roc_auc값, classfication_report를 계산하시오
from sklearn.metrics import accuracy_score
from sklearn.metrics import roc_auc_score, f1_score
from sklearn.metrics import classification_report # 분류 성능 평가 보고서(macro avg 사용)

# accuracy 
print(accuracy_score(y_test,Y_pred))

# roc_auc
print(f1_score(y_test,Y_pred))

# classification_report
print(classification_report(y_test,Y_pred))
```

```python
# 분류 예측갑소가 분류 확률 값을 계산하여 index 제외하고 csv ㅏ파일에 저장하여 제출 하시오
pred_label = model.predict(x_test1)
pred_label.astype('int64') # 문제에서 데이터 타입 정하는 경우

pred_proba = model.predict_proba(x_test1)[:,1]
print(pred_proba)
# pred_proba = np.round(pred_proba,2) ## 결과 소수점 2자리로 표시해야 할 때

pd.DataFrame({'분류예측':pred_label,'분류확률':pred_proba}).to_csv('1234_results.csv',index=False)
```
